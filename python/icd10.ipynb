```python
def format_icd10(code):
    code = code.strip()
    if len(code) > 3:
        return code[:3] + "." + code[3:]
    return code
```

```python
rows = []
with open("E:\Projects\SQL\Healthcare Project\HealthCare Data\icd10cm_codes_2026.txt", "r", encoding="latin-1") as f:
    for line in f:
        if not line.strip():
            continue
        raw_code = line[0:7].strip()
        description = line[8:].strip()
        rows.append((format_icd10(raw_code), description))

df_dot = pd.DataFrame(rows, columns=["diagnosis_code", "description"])
```
